```python
!pip install nltk

import nltk
from nltk.chat.util import Chat, reflections

patterns = [
    [
        r"hi|hello|hey",
        ["Hello! I am the Plant Health Assistant. How can I help?"]
    ],
    [
        r".yellow leaves.",
        ["Yellow leaves may indicate overwatering, lack of nutrients, or disease. Try checking soil moisture and leaf spots."]
    ],
    [
        r".brown spots.",
        ["Brown spots can be caused by fungal diseases, sunburn, or nutrient problems. Upload an image for better analysis."]
    ],
    [
        r".water.*tomato.",
        ["Tomato plants usually need regular watering, but the soil should not stay too wet."]
    ],
    [
        r".temperature.",
        ["Tomato plants grow best in warm temperatures, usually around 20-30°C."]
    ],
    [
        r".upload.*image.|.photo.",
        ["You can upload a plant image, and the system will analyze if it is a tomato plant and check possible disease symptoms."]
    ],
    [
        r".project.",
        ["This project uses AI and IoT ideas to help monitor tomato plant health."]
    ],
    [
        r"bye|exit|quit",
        ["Goodbye! Take care of your plants."]
    ]
]

chatbot = Chat(patterns, reflections)

print("Plant Chatbot is ready! Type 'bye' to exit.")

while True:
    user_input = input("You: ")
    if user_input.lower() in ["bye", "exit", "quit"]:
        print("Bot: Goodbye!")
        break

    response = chatbot.respond(user_input)
    if response:
        print("Bot:", response)
    else:
        print("Bot: I am not sure. Try asking about leaves, watering, temperature, or plant diseases.")
```
